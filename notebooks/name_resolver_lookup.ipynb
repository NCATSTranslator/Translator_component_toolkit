# Using `name_resolver` and `node_normalizer`

```python
from TCT import name_resolver, node_normalizer
```

## `name_resolver`

This module can be used to look up nodes and CURIEs given names of biological objects.

### `lookup`

```python
name_resolver.lookup('TP53')
```

```python
name_resolver.lookup('Acute Myeloid Leukemia')
```

```python
name_resolver.lookup('COVID-19')
```

### Query parameters

- `only_taxa`: Requiring specific taxa (species) - for example, human is 'NCBITaxon:9606'
- `biolink_type`: Selecting specific node types - 'biolink:Disease', 'biolink:Gene', etc.
- `return_top_response` and `limit` - whether to return the top response only, and how many responses to include.

```python
name_resolver.lookup('AML', biolink_type='biolink:Disease')
```

```python
name_resolver.lookup('AML', biolink_type='biolink:Disease', return_top_response=False)
```

```python
name_resolver.lookup('P53', only_taxa='NCBITaxon:9606', biolink_type='biolink:Gene')
```

### Batch lookup

With `batch_lookup`, we can look up multiple names at the same time. Many of the query parameters for `lookup` can also be used here, such as `only_taxon`, but instead of `biolink_type`, we have `biolink_types`, a list of biolink types.

```python
name_resolver.batch_lookup(['KRAS', 'NRAS'], only_taxa='NCBITaxon:9606')
```

```python
name_resolver.batch_lookup(['P53', 'TP53', 'KRAS', 'NRAS', 'NPM1'], only_taxa='NCBITaxon:9606', biolink_types=['biolink:Gene'])
```

## `node_normalizer`

This module is used to get names for CURIE ids, as well as get node objects given CURIEs.

### `get_preferred_names`

This returns a dict that maps CURIE ids to preferred (human-readable) names.

```python
output = node_normalizer.get_preferred_names(['NCBIGene:3458', 'MONDO:0100096', 'NCBIGene:1055', 'NCBIGene:28755'])
```

```python
output
```

```python
node_normalizer.get_preferred_names(['NCBIGene:3458', 'MONDO:0100096', 'NCBIGene:1055', 'NCBIGene:28755', 'afds;lkdasfjdf;l'])
```

### `get_normalized_nodes`

This returns a dict that maps CURIE ids to nodes.

```python
output = node_normalizer.get_normalized_nodes(['NCBIGene:3458', 'MONDO:0100096', 'NCBIGene:1055', 'NCBIGene:28755'])
```

```python
output
```

```python
output = node_normalizer.get_normalized_nodes(['NCBIGene:3458', 'MONDO:0100096', 'NCBIGene:1055', 'NCBIGene:28755'], mode='post')
```
