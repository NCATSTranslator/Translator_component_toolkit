```python
from TCT import translator_metakg, translator_query, TCT
import pandas as pd
```

```python
# Step 1: Load Translator resources (APIs, metaKG)
from TCT.translator_resources import TranslatorResources
resources = TranslatorResources.load()

print(f"Number of APIs: {len(resources.api_names)}")
print(f"Number of predicates: {len(list(set(resources.meta_kg['Predicate'])))}")
```

```python
input_node_list = [TCT.get_curie('NPM1'), TCT.get_curie('FLT3'), TCT.get_curie('NRAS'), TCT.get_curie('BCL2')]
```

```python
input_node1_category = ['biolink:Gene'] # Node: this has to be in a format of biolink:xxx
input_node2_category = ['biolink:Gene']
sele_predicates = list(set(TCT.select_concept(sub_list=input_node1_category, obj_list=input_node2_category, metaKG=resources.meta_kg)))
sele_APIs = TCT.select_API(sub_list=input_node1_category, obj_list=input_node2_category, metaKG=resources.meta_kg)
API_URLs = TCT.get_Translator_API_URL(API_sele=sele_APIs, APInames=resources.api_names)
```

```python
query_json = TCT.format_query_json(subject_ids=input_node_list,
                                   object_ids=[],
                                   subject_categories=input_node1_category,
                                   object_categories=input_node2_category,
                                   predicates=sele_predicates)
```

```python
# Step 4: Query APIs in parallel and parse annotations
result = translator_query.parallel_api_query(query_json=query_json,
                                             select_APIs=list(sele_APIs),
                                             resources=resources,
                                             max_workers=len(sele_APIs))
pairs_found = TCT.get_pair_annotation(result=result, input_node_list=input_node_list)
edge_list = TCT.parse_pair_annotation(pairs_found=pairs_found, input_node_list=input_node_list)
```
