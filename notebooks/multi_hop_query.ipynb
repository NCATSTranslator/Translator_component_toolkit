# Multi-hop TRAPI Queries

This notebook demonstrates how to build multi-hop TRAPI query graphs using `HopSpec` and `build_multi_hop_query()`.

```python
import json

from TCT.trapi import HopSpec, build_multi_hop_query, build_query
```

## Single-hop query

A single `HopSpec` produces the same structure as the original `build_query()` function.

```python
query = build_multi_hop_query(
    subject_ids=["NCBIGene:3845"],
    hops=[
        HopSpec(
            predicates=["biolink:physically_interacts_with"],
            object_categories=["biolink:Gene"],
        )
    ],
    return_json=False,
)
print(json.dumps(query, indent=2))
```

## Two-hop query

Chain two hops to find paths like Gene -> BiologicalProcess -> Disease.

```python
query = build_multi_hop_query(
    subject_ids=["NCBIGene:3845"],
    subject_categories=["biolink:Gene"],
    hops=[
        HopSpec(
            predicates=["biolink:actively_involved_in"],
            object_categories=["biolink:BiologicalProcess"],
        ),
        HopSpec(
            predicates=["biolink:associated_with"],
            object_ids=["MONDO:0018874"],
        ),
    ],
    return_json=False,
)
print(json.dumps(query, indent=2))
```

## Three-hop chain

Longer chains work the same way. Each `HopSpec` adds one edge and one node to the query graph.

```python
query = build_multi_hop_query(
    subject_ids=["MONDO:0018874"],
    subject_categories=["biolink:Disease"],
    hops=[
        HopSpec(
            predicates=["biolink:has_phenotype"],
            object_categories=["biolink:PhenotypicFeature"],
        ),
        HopSpec(
            predicates=["biolink:associated_with"],
            object_categories=["biolink:Gene"],
        ),
        HopSpec(
            predicates=["biolink:physically_interacts_with"],
            object_categories=["biolink:ChemicalEntity"],
        ),
    ],
    return_json=False,
)
print(json.dumps(query, indent=2))
```

## Omitting optional fields

When predicates or categories are `None`, they are cleanly omitted from the query graph.

```python
query = build_multi_hop_query(
    subject_ids=["NCBIGene:7157"],
    hops=[
        HopSpec(object_categories=["biolink:Disease"]),
        HopSpec(object_categories=["biolink:ChemicalEntity"]),
    ],
    return_json=False,
)
print(json.dumps(query, indent=2))
```
